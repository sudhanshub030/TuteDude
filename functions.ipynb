### TASK 1: Basic Function Price After Discount


```python
def apply_discount(price, discount_percent=5):
    
    if discount_percent > 60:
        discount_percent = 60
    
    discount_amount = price * (discount_percent / 100)
    
    final_price = price - discount_amount
    
    return final_price


#function
print(apply_discount(1000, 10))
print(apply_discount(500))
```

### TASK 2 : Recursive Function : Factorial Utility

```python
def factorial(n):
    if n < 0:
        print("Error.Factorial is not defined for negative numbers.")
        return None
    
    elif n == 0 or n == 1:
        return 1
    
    else:
        
        return n * factorial(n-1)
    
print(factorial(5))
print(factorial(0))
print(factorial(-1))
```

### TASK 3 : Lambda Function : GST Calculation

```python
gst = lambda price: price + (0.18 * price)

print(gst(100))

final_price = lambda price, discount = 10 : (price + 0.18*price) - ((price + 0.18*price)*discount/100)

print(final_price(100))
```

### TASK 4 : Using map(): Apply GST to List Price

```python
prices = [100,250,400,1200,50]

gst_rate = 0.18

prices_with_gst = list(map(lambda price: price + (price * gst_rate), prices))

print("Original prices:", prices)
print("Prices after GST:", prices_with_gst)
```

### Task 5: Task 5 – Using filter()

```python
prices = [100, 250, 400, 1200, 50, 2000, 850]

greater_than_500 = list(filter(lambda x: x > 500, prices))

less_equal_500 = list(filter(lambda x: x <= 500, prices))

print("Prices > 500:", greater_than_500)
print("Prices <= 500:", less_equal_500)
```

### Task 6 – Combined Utility Function

```python
def process_prices(prices):
    discounted_prices = list(map(lambda x: x * 0.9, prices))
    
    filtered_prices = list(filter(lambda x: x > 300, discounted_prices))
    
    return discounted_prices, filtered_prices

result = process_prices([100, 500, 900, 50, 750])

print("Discounted Prices:", result[0])
print("Filtered Prices (>300):", result[1])
```

### Task 7 – Mini Menu Using Functions

```python
# Function to add price
def add_price(prices_list, price):
    prices_list.append(price)

# Function to get average price
def get_average_price(prices_list):
    if len(prices_list) == 0:
        return 0
    return sum(prices_list) / len(prices_list)

# Function to get max price
def get_max_price(prices_list):
    if len(prices_list) == 0:
        return None
    return max(prices_list)


# Main Menu
prices = []

while True:
    print("\nMenu:")
    print("1 → Add price")
    print("2 → Show average price")
    print("3 → Show highest price")
    print("q → Quit")

    choice = input("Enter choice: ")

    if choice == '1':
        price = float(input("Enter price: "))
        add_price(prices, price)

    elif choice == '2':
        print("Average Price:", get_average_price(prices))

    elif choice == '3':
        print("Highest Price:", get_max_price(prices))

    elif choice == 'q':
        print("Exiting program...")
        break

    else:
        print("Invalid choice!")
```
